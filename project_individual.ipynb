3) Exploratory Data Analysis and Visualization
To load the dataset into Python, following command will be used:

```python
import pandas as pd
players = pd.read_csv("https://drive.google.com/uc?export=download&id=1Mw9vW0hjTJwRWx0bDXiSpYsO3gKogaPz")
players
```

Minimum necessary wrangling:

```python
players_clean = players.iloc[:, :7]
players_clean

players_clean["hashedEmail"].str.split("/", expand=True)
players_tidy = players_clean.drop(columns=["hashedEmail"])
players_tidy
```

Explanatory visualizations:

```python
import altair as alt
players_hrs_age = alt.Chart(players_tidy).mark_circle(opacity=0.4).encode(x=alt.X("age").title("Player's Age"), y=alt.Y("played_hours").title("Played Hours"))
players_hrs_age

players_hrs_gender = alt.Chart(players_tidy).mark_circle(opacity=0.4).encode(x=alt.X("gender").title("Player's Gender"), y=alt.Y("played_hours").title("Played Hours"))
players_hrs_gender

players_hrs_exp = alt.Chart(players_tidy).mark_circle(opacity=0.4).encode(x=alt.X("experience").title("Player's level"), y=alt.Y("played_hours").title("Played Hours"))
players_hrs_exp
```

From this visualizations, we know that the data varies so much that when using scatter plot, the data is not comparable unless additional code is used to standardize the data, such as log.
